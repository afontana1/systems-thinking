# Lattice Boltzmann (D2Q9)

Runs the LBM simulator and saves velocity snapshots.

```python
import importlib.util
from pathlib import Path

module_path = Path("Lattice Boltzmann.py")
spec = importlib.util.spec_from_file_location("lbm", module_path)
lbm = importlib.util.module_from_spec(spec)
spec.loader.exec_module(lbm)

lbm.main()
```
