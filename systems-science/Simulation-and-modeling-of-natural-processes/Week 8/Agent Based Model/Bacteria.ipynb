# Agent Based Model - Bacteria Chemotaxis

Runs the bacteria simulation with live visualization.

```python
import importlib.util
from pathlib import Path

module_path = Path("Bacteria behaviour.py")
spec = importlib.util.spec_from_file_location("bacteria", module_path)
bacteria = importlib.util.module_from_spec(spec)
spec.loader.exec_module(bacteria)

bacteria.main()
```
