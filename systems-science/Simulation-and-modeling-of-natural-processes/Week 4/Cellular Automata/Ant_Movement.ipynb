# Cellular Automata - Langton's Ant

This notebook runs the Langton's Ant simulation using the updated Python 3.12 code.

```python
import importlib.util
from pathlib import Path

module_path = Path("Automata Ant.py")
spec = importlib.util.spec_from_file_location("automata_ant", module_path)
automata_ant = importlib.util.module_from_spec(spec)
spec.loader.exec_module(automata_ant)

automata_ant.main()
```
